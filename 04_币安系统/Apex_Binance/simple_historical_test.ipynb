## 📈 1. 获取历史数据

```python
def fetch_historical_data(symbol='BTC/USDT', timeframe='1h', days=7):
    """获取历史数据"""
    try:
        from core.exchange_client import exchange_client
        
        # 初始化模拟模式
        exchange_client.initialize(demo_mode=True)
        
        print(f"📊 获取 {symbol} {timeframe} 数据...")
        
        # 获取数据
        ohlcv = exchange_client.fetch_ohlcv(symbol, timeframe, limit=100)
        
        if ohlcv is not None and not ohlcv.empty:
            df = ohlcv.copy()
            print(f"✅ 成功获取 {len(df)} 条数据")
            return df
        else:
            print(f"❌ 数据获取失败")
            return None
            
    except Exception as e:
        print(f"❌ 错误: {e}")
        return None

# 测试获取数据
btc_data = fetch_historical_data('BTC/USDT', '1h', 7)
if btc_data is not None:
    print(f"\n📋 数据概览:")
    print(f"   数据形状: {btc_data.shape}")
    print(f"   时间范围: {btc_data.index[0]} 到 {btc_data.index[-1]}")
    print(f"   价格范围: {btc_data['close'].min():.2f} - {btc_data['close'].max():.2f}")
```

## 📊 2. 查看数据

```python
if btc_data is not None:
    print("📊 前5条数据:")
    print(btc_data.head())
    
    print("\n📊 后5条数据:")
    print(btc_data.tail())
    
    print("\n📊 基本统计:")
    print(btc_data[['open', 'high', 'low', 'close', 'volume']].describe())
```

## 📈 3. 简单可视化

```python
if btc_data is not None:
    try:
        # 创建图表
        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 8))
        
        # 价格图表
        ax1.plot(btc_data.index, btc_data['close'], label='收盘价', color='blue', linewidth=1)
        ax1.set_title('BTC/USDT 价格走势')
        ax1.set_ylabel('价格 (USDT)')
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        
        # 成交量图表
        ax2.bar(btc_data.index, btc_data['volume'], color='green', alpha=0.6)
        ax2.set_title('成交量')
        ax2.set_ylabel('成交量')
        ax2.grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()
        
        print("✅ 图表生成成功")
        
    except Exception as e:
        print(f"❌ 图表生成失败: {e}")
```
